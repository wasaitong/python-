### 构造时间序列date_range\bdate_range() 


- 实际工作中，经常要生成含大量时间戳的超长索引，一个个输入时间戳又枯燥，又低效
- 如果时间戳是定频的，用date_range()与bdate_range()函数即可创DatetimeIndex。
- date_range 默认的频率是日历日，bdate_range 的默认频率是工作日

```python
pd.date_range(start='2008-01-01', end='2020-01-01', freq='2Y', name='时间')
```

```python
pd.bdate_range(start='2022-02-01', end='2022-02-16', freq='2B')
```

```python
# periods 指定时间序列的数据个数，注意不要再指定end结束时间，否则可能会产生冲突
pd.bdate_range(start='2022-02-16' ,periods=5, freq='2B')
```

### 无效数据

不可解析时，默认值 errors='raise' 会触发错误：

```python
# 有错就要处理，默认值就Ok
pd.to_datetime(['2009/07/30', 'asd'], errors='raise')
```

errors='ignore' 返回原始输入：

```python
indexes = pd.to_datetime(['2009-07-30', 'asd'], errors='ignore')
```

```python
indexes[0]
```

errors='coerce' 把无法解析的数据转换为 NaT，即不是时间（Not a Time）：

```python
indexes = pd.to_datetime(['2009-07-30', 'asd'], errors='coerce')
```

```python
indexes[0]
```

```python
pd.to_datetime('2012年12月11日', errors='ignore')
```

```python
def my_transform(x):
    if x == None:
        return None
    
    # 解决'2012年12月11日'类问题
    if x.endswith('日'):
        return pd.to_datetime(x, format='%Y年%m月%d日')
    
    # 解决'2012-01-01', 'asd'两类
    obj = pd.to_datetime(x, errors='ignore')
    
    if isinstance(obj, str):
        return np.nan
    else:
        return obj
```

```python
pd.Series(data=['2012-01-01', 'asd', '2012年12月11日', '2003-02-03', None]).map(my_transform)
```

## 重采样

Pandas 有一个虽然简单，但却强大、高效的功能，可在频率转换时执行重采样，如，将秒数据转换为 5 分钟数据，这种操作在金融等领域里的应用非常广泛。

resample() 是基于时间的分组操作，每个组都遵循归纳方法。

从 0.18.0 版开始，resample() 可以直接用于 DataFrameGroupBy 对象，

注意：重采样一定要以时间类型做Index

```python
DatetimeIndex
```

```python
AAPL['year'] = AAPL['Date'].dt.year
```

```python
AAPL.head()
```

```python
AAPL.groupby('year')['Open'].mean()
```

```python
# 【前提】 必须要用时间作为索引，否则不可以进行重采样操作
temp = AAPL.set_index('Date')
```

```python
# 分组必聚合
temp.resample('Y')['Open'].mean()
```

```python
# 每3年做一次平均开盘价的计算
temp.resample('3Y')['Open'].mean()
```

```python
temp.resample('450D')['Open'].mean()
```

resample 函数非常灵活，可以指定多种频率转换与重采样参数。

支持聚合函数的处理如： sum、mean、std、sem、max、min、mid、median

## 时间差

Timedelta() 支持用多种参数生成时间差：

```python
now = pd.datetime.now()
now
```

```python
delta = pd.Timedelta(4, unit='day')
delta
```

```python
future = now + delta
```

```python
future - now
```

```python
delta = pd.Timedelta(days=1, seconds=1)
```

```python
now + delta
```

以时间差为数据的 Series 与 DataFrame 支持各种运算，datetime64 [ns] 序列或 Timestamps 减法运算生成的是timedelta64 [ns] 序列。

```python
s = pd.Series(pd.date_range('2012-1-1', periods=3, freq='D'))
td = pd.Series([pd.Timedelta(days=i) for i in range(3)])
df = pd.DataFrame({
    'A':s,
    'B':td
})
df
```

```python
df['C'] = df['A'] + df['B']
```
